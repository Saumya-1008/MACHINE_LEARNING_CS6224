```python
import numpy as np

print("=== Simple Linear Regression ===")

# 1. Take dynamic inputs for X and y
x_str = input("Enter X values separated by space (e.g., 1 2 3 4 5): ")
X = np.array([float(i) for i in x_str.strip().split()])

y_str = input("Enter y values separated by space (e.g., 45 50 55 65 70): ")
y = np.array([float(i) for i in y_str.strip().split()])

if len(X) != len(y):
    print("Error: X and y must contain the same number of data points.")
else:
    # 2. Hyperparameters
    learning_rate = 0.01 
    epochs = 5000
    m, c = 0.0, 0.0
    n = len(X)

    # 3. Gradient Descent Loop
    for _ in range(epochs):
        y_pred = m * X + c
        D_m = (-2 / n) * np.sum(X * (y - y_pred))
        D_c = (-2 / n) * np.sum(y - y_pred)
        m -= learning_rate * D_m
        c -= learning_rate * D_c

    print(f"\nModel trained! Equation: y = {m:.4f}X + {c:.4f}")

    # 4. Predict for a user-provided value
    val_str = input("\nEnter an X value to predict its corresponding y: ")
    target_x = float(val_str.strip())
    predicted_y = m * target_x + c

    print(f"Predicted value for X={target_x}: {predicted_y:.4f}")
```
