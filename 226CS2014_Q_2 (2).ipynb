# 2. Multivariate regression: Build a model to predict house price using at least three features (e.g., area, bedrooms, age). Implement fit(), predict(), MSE, MAE, and R² without sklearn regression.

```python
import numpy as np

class MultivariateLinearRegression:
    def __init__(self):
        self.weights = None
        
    def fit(self, X, y):
        # Add bias column (1s) to X
        X_b = np.c_[np.ones((X.shape[0], 1)), X]
        # Normal Equation: theta = (X^T * X)^-1 * X^T * y
        self.weights = np.linalg.inv(X_b.T.dot(X_b)).dot(X_b.T).dot(y)
        
    def predict(self, X):
        X_b = np.c_[np.ones((X.shape[0], 1)), X]
        return X_b.dot(self.weights)
    
    def mse(self, y_true, y_pred):
        return np.mean((y_true - y_pred) ** 2)
    
    def mae(self, y_true, y_pred):
        return np.mean(np.abs(y_true - y_pred))
    
    def r2(self, y_true, y_pred):
        ss_residual = np.sum((y_true - y_pred) ** 2)
        ss_total = np.sum((y_true - np.mean(y_true)) ** 2)
        return 1 - (ss_residual / ss_total)

print("=== Multivariate Linear Regression ===")

# 1. Define dataset dimensions dynamically
n_samples = int(input("Enter total number of samples (rows): "))
n_features = int(input("Enter number of features (e.g., 3 for Area, Beds, Age): "))

X_list = []
y_list = []

print(f"\nProvide data for {n_samples} samples.")
print(f"For each sample, enter {n_features} features followed by the target value, all separated by spaces.")

# 2. Collect dataset dynamically
for i in range(n_samples):
    row_input = input(f"Sample {i+1} ({n_features} features + 1 target): ")
    values = [float(val) for val in row_input.strip().split()]
    
    # Split features and target
    X_list.append(values[:-1])
    y_list.append(values[-1])

X_train = np.array(X_list)
y_train = np.array(y_list)

# 3. Train the model
model = MultivariateLinearRegression()
model.fit(X_train, y_train)
predictions = model.predict(X_train)

# 4. Evaluate on the inputted data
print("\n--- Training Results ---")
print(f"Weights (Intercept, Feature1, ...): {np.round(model.weights, 4)}")
print(f"MSE: {model.mse(y_train, predictions):.4f}")
print(f"MAE: {model.mae(y_train, predictions):.4f}")
print(f"R² Score: {model.r2(y_train, predictions):.4f}")

# 5. Predict on completely new, unseen features
print("\n--- Make a New Prediction ---")
new_input = input(f"Enter {n_features} features separated by space to predict the target: ")
new_X = np.array([[float(val) for val in new_input.strip().split()]])

new_prediction = model.predict(new_X)
print(f"Predicted target value: {new_prediction[0]:.4f}")
```
